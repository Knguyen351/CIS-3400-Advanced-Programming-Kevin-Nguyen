# Calculate Final Grade

This notebook defines `calculate_final_grade()`, which drops the lowest quiz grade from a list of numeric grades and returns the average of the remaining grades.

No imports are needed — only Python built-ins (`sorted`, `sum`, `len`) are used.

```python
def calculate_final_grade(grades):
    """
    Drops the lowest grade from a list of quiz grades and returns
    the average of the remaining grades.
    """
    if len(grades) <= 1:
        raise ValueError("Need at least 2 grades to drop the lowest and average the rest.")

    remaining_grades = sorted(grades)[1:]  # remove the lowest grade
    return sum(remaining_grades) / len(remaining_grades)
```

```python
grades = [85, 92, 78, 90, 88]
print(calculate_final_grade(grades))
```

```python
# Standard midpoint values used to convert letter grades to numeric scores
LETTER_GRADE_SCALE = {
    'A': 95,
    'B': 85,
    'C': 75,
    'D': 65,
    'F': 50,
}


def calculate_final_grade(grades):
    """
    Drops the lowest grade from a list of quiz grades and returns
    the average of the remaining grades.

    Accepts numeric grades (int/float, 0-100) and/or letter grades
    ('A', 'B', 'C', 'D', 'F', case-insensitive). Letter grades are
    converted to numeric equivalents before averaging.

    Edge cases:
    - Empty list -> returns 0
    - Single grade -> returns that grade as-is (nothing to drop)
    """
    if len(grades) == 0:
        return 0

    numeric_grades = []
    for grade in grades:
        if isinstance(grade, str):
            letter = grade.strip().upper()
            if letter not in LETTER_GRADE_SCALE:
                raise ValueError(f"Invalid letter grade: {grade!r}. Must be one of A, B, C, D, F.")
            numeric_grades.append(LETTER_GRADE_SCALE[letter])
        elif isinstance(grade, (int, float)):
            if not (0 <= grade <= 100):
                raise ValueError(f"Numeric grade out of range: {grade}. Must be between 0 and 100.")
            numeric_grades.append(grade)
        else:
            raise TypeError(f"Unsupported grade type: {type(grade).__name__}. Use a number or a letter grade string.")

    if len(numeric_grades) == 1:
        return numeric_grades[0]

    remaining_grades = sorted(numeric_grades)[1:]  # remove the lowest grade
    return sum(remaining_grades) / len(remaining_grades)


grades = [85, 'B', 'A', 92, 'C']
print(calculate_final_grade(grades))


grades=[]
print(calculate_final_grade(grades))

grades=[85]
print(calculate_final_grade(grades))
```
